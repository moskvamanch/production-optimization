```python
import random
import matplotlib.pyplot as plt

def exp_time(mean):
    return random.expovariate(1 / mean)

times = [exp_time(10) for _ in range(10000)]

plt.figure(figsize=(8, 5))
plt.hist(times, bins=50)
plt.xlabel("Processing time, minutes")
plt.ylabel("Frequency")
plt.title("Exponential processing times with mean = 10 minutes")
plt.grid(True)
plt.show()

print("Average:", sum(times) / len(times))
```
